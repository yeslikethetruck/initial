```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

plt.rcParams["figure.figsize"] = (15,10)

d = stats.norm.rvs(loc = 7.0, scale = .05, size = 1000000) # our "random variates" normal distribution courtesy of scipy

 

fig, ax = plt.subplots(1, 1) # a classic, really
ax.hist(d,75, density=True, color = 'goldenrod') # picking a nice color for our histogram
plt.tick_params(labelsize = 24) 
plt.xlim([6.75,7.25]) # centered around our new point
x = np.linspace(6.75,7.25,1000) # creating the co-ordinate space to call our functions with
ax.plot(x,stats.norm.pdf(x,loc = 7, scale = 0.05),linewidth = 8,alpha = 0.7, color = 'firebrick', label = r'$ f(x) = \frac{1}{\sqrt{2\pi}}e^\frac{-(x-7)^2}{2} $') # non default color, ax.plot taking an x and y
ax.legend(fontsize = 20) # so we can see that the standard gaussian formula does indeed fit, the standard deviation is one from our distribution
plt.show() # necessary
```

```python
ax = plt.hist(d,75, density=True, color = 'goldenrod') # unexpected that this is necessary to repeat
plt.yscale('log') # now we look at the y-log plot to see if it's really gaussian ...
plt.tick_params(labelsize = 24)
plt.xlim([6.75,7.25])
plt.show()
# should look like a parabola and it does (with some minor outliers [i presume] visible)
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,75, density=True, color = 'firebrick')
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([6.75,7.25])
x = np.linspace(6.75,7.25,1000)
ax.plot(x,stats.norm.pdf(x,loc = 7, scale = 0.05),linewidth = 8,alpha = 0.7, color = 'goldenrod', label = r'$ f(x) = \frac{1}{\sqrt{2\pi}}e^\frac{-(x-7)^2}{2} $') # let's mix up those colors
# now our gaussian fit appears parabolic in the log scale, which is expected due to the argument of the exponential
ax.legend(fontsize = 20)
plt.show()
```
